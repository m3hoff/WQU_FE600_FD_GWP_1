Data Quality

#### 1.a) Example of poor-quality structured data:

An example of poor-quality structured data is the dataset for Request for Trade (RFQ) messages for Interest Rate Swaps (IRS). IRS is an over-the-counter (OTC) product with no standardized trade validation by legal bodies like stock exchanges. As a result, RFQ datasets often have the following issues:

Key fields may be left blank, such as notional amounts or payment conventions, due to incomplete information provided by clients or dealers.
Some fields may contain inaccurate or untrustworthy information because OTC transactions allow for post-trade amendments under mutual agreement.
Data may be improperly categorized due to unenumerable contents like custom coupon conventions or reference rate agreements that deviate from typical specifications.
Critical information like deal structures may be highly human-customized, leading to inconsistencies or ambiguities in the dataset.


#### 1.b) Recognizing poor-quality structured data:

Structured data is considered of poor quality when it fails to meet key properties of good-quality data. For example:

- Non-existence: Fields may lack values entirely, leaving critical cells blank.
- Lack of trust: Values provided are either inaccurate or cannot be trusted due to system or user errors.
- Unenumerability: Some fields contain values that cannot be categorized or quantified due to non-standardized conventions.
- Meaninglessness: Even if the data is filled, it may not lead to meaningful insights, such as cases where values are scattered too diversely for statistical analysis.

For instance, stock transaction datasets from exchanges like the London Stock Exchange are reliable, as they are monitored under strict regulatory frameworks and adhere to enumerated conventions. This contrasts sharply with the inconsistency and ambiguity seen in OTC RFQ data.


#### 1.c) Example of poor-quality unstructured data:

An example of poor-quality unstructured data is Bloomberg broker chat messages. These chatrooms often contain free-form text written by traders, discussing themes like FX swaps or IRS options. When gathered for analytics, the chats exhibit the following issues:

Unstructured sentences with heavy jargon and abbreviations, making parsing and categorization challenging.
Lack of standardized formats, as traders are free to write anything, leading to inconsistency and noise.
High dependency on natural language processing to extract actionable insights, with a typically low success rate due to fragmented and noisy data.


#### 1.d) Recognizing poor-quality unstructured data:

Unstructured data can be even harder to assess compared to structured data due to the following reasons:

- Quality: Like structured data, unstructured data may lack trustworthiness. For example, traders can leave irrelevant or nonsensical messages without any repercussions.
- Noise: Unrelated or irrelevant chatlines, such as personal conversations or accidental messages in the wrong chatroom, can pollute the data and reduce its usefulness.
- Accessibility: Access to unstructured data like broker chats often requires permission from providers. This access can be limited or costly and may not guarantee long-term availability.
- Specification: The lack of predefined fields and topics in unstructured data leads to vast variability in what analysts need to parse, making the data extraction process ambiguous and prone to errors.

In contrast, Federal Reserve meeting minutes are an example of high-quality unstructured data. They are consistently stated with key fields provided, publicly accessible, and provide clear objectives, enabling analysts to focus on extracting relevant information regarding monetary policies.
